```python
import pandas as pd
from sklearn import tree
import joblib


music_data=pd.read_csv('music-extended.csv.')
print(music_data.shape)
music_data.describe()
music_data.info()
```

```python
X=music_data.drop(columns=['genre'])
y=music_data['genre']




model = tree.DecisionTreeClassifier()
model.fit(X,y)
```

```python
joblib.dump(model,'saved-music-recommender.joblib')
```
