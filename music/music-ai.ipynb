```python
import pandas as pd
from sklearn import tree 



music_data=pd.read_csv('music-extended.csv.')
#print(music_data.shape)
print("---------shape----------------")
#print(music_data.describe())
print("----------describe---------------")

music_data.info()
```

```python
X=music_data.drop(columns=['genre'])
y=music_data['genre']

print(y.unique())
print(y.value_counts())


model = tree.DecisionTreeClassifier()
model.fit(X,y)

predictions=model.predict([ [21,1], [22,0]])
#predictions=model.predict([[82,1]])
predictions
```

```python
predictions=model.predict([[103,1]])
predictions
```

```python

print(type(y))
print(y.unique())
    

tree.export_graphviz(model, out_file='ai-music-recommender.dot',
                     feature_names=['age','gender'],
                     class_names=sorted(y.unique()),
                     label='all',
                     rounded=True,
                     filled=True)
```
