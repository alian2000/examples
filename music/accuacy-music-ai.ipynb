```python
import pandas as pd
from sklearn import tree
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt


music_data=pd.read_csv('music-extended.csv.')
print(music_data.shape)
music_data.describe()
music_data.info()
```

```python
X=music_data.drop(columns=['genre'])
y=music_data['genre']
#accuracy
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.9)



model = tree.DecisionTreeClassifier()
model.fit(X_train,y_train)
predictions=model.predict(X_test)
score=accuracy_score(y_test,predictions)
print(len(y_test),len(predictions))
plt.scatter(y_test,predictions)
plt.xlabel('Actual Genre Width')
plt.ylabel('Predicted Genre')
plt.title('Actual vs Predicted Genre')

plt.show()
score
predictions=model.predict([ [21,1], [22,0]])
predictions=model.predict([[40,1]])
print("_____________________________________________________________________________________________________________")
print("actual value= ",y_test)
predictions
print("_____________________________________________________________________________________________________________")
```
