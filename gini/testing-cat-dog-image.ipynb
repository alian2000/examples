```python
import os
import cv2
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
import joblib
```

```python
clf = joblib.load('dog_vs_cat_decision_tree.pkl')
print("Model loaded successfully!")
```

```python
data_raw1=[]
img1 = cv2.imread("noor-cat-01.jpg")
IMG_SIZE=64
img1 = cv2.resize(img1, (IMG_SIZE, IMG_SIZE))
plt.imshow(img1)
plt.show()
        
# Flatten the image into a vector and append to data
data_raw1.append(img1.flatten())
dataf=np.array(data_raw1).reshape(1,-1)
print("prediction= ",clf.predict(dataf))
```
